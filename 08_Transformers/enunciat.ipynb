## Les tres famílies de Transformers

| Arquitectura | Com funciona | Per a què serveix | Exemples de models |
|---|---|---|---|
| **Només encoder** | Llegeix tota l'entrada en totes dues direccions i en produeix una representació | **Entendre** text: classificació, sentiment, entitats amb nom (NER), resposta extractiva a preguntes | BERT, RoBERTa, DistilBERT |
| **Només decoder** | Prediu el token següent d'esquerra a dreta | **Generar** text: completar frases, xat, codi | GPT-2, Llama, Mistral |
| **Encoder–decoder (Transformer original)** | L'encoder llegeix l'entrada i el decoder genera una seqüència nova | **Transformar** una seqüència en una altra: traducció, resum, paràfrasi | T5, BART, MarianMT |
